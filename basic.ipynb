### 17. Parcours
**En largeur — BFS**

À connaître pour trouver un plus court chemin dans une grille ou un graphe dont les arêtes ont toutes le même coût

```python
from collections import deque

def bfs(graphe, depart, arrivee):
    file = deque([depart])
    distances = {depart: 0}

    while file:
        sommet = file.popleft()

        if sommet == arrivee:
            return distances[sommet]

        for voisin in graphe.get(sommet, []):
            if voisin not in distances:
                distances[voisin] = distances[sommet] + 1
                file.append(voisin)

    return None
```

**En profondeur — DFS**

Utile pour explorer une zone, tester l’accessibilité ou identifier des composantes connexes

```python
pile = [depart]
visites = {depart}

while pile:
    sommet = pile.pop()

    for voisin in graphe.get(sommet, []):
        if voisin not in visites:
            visites.add(voisin)
            pile.append(voisin)
```

version itérative évite la limite de profondeur de récursion de Python.
